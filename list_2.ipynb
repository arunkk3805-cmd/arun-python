```python
nums=[21,55,88,44,66,3,4]
nums.sort()
print(nums)
nums.sort(reverse=True)
print(nums)
```

shallow copy

```python
a=[1,2,3,4,5]
print(a)
c=a
c.insert(1,1223)
print(c)
print(a)
b=a.copy()
b.append("*")
print(a)
print(b)
```
